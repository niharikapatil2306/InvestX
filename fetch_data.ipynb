```python
import yfinance as yf
import pandas as pd
import os
from datetime import datetime, timedelta
import time
```

```python
def load_companies_from_csv(csv_file):
    df = pd.read_csv(csv_file, sep=';')
    companies = []
    
    for _, row in df.iterrows():
        companies.append({
            'Symbol': row['Symbol'],
            'Company_Name': row['Company_Name'],
            'Market_Cap': row['Market_Cap'],
            'Price': row['Price'],
            'Change_%': row['Change_%'],
            'Sector': row['Sector'],
        })
    
    return companies
```

```python
def fetch_data(symbol):
    symbols_to_try = [symbol]
    if not symbol.endswith('.L'):
        symbols_to_try.append(symbol + '.L')
        
    for s in symbols_to_try:
        print(s)
        ticker = yf.Ticker(s)
        hist_data = ticker.history(period="5y")
            
        if not hist_data.empty:
            return {
                'symbol_used': s,
                'historical': hist_data,
                'info': ticker.info,
                'financials': ticker.financials,
                'balance_sheet': ticker.balance_sheet,
                'cashflow': ticker.cashflow
                }
        else:
            continue
    return None
```

```python
def save_data(data, symbol, output_folder="data"):
    if data is None:
        return

    actual_symbol = data.get('symbol_used', symbol)
    out_dir = f'{output_folder}/{actual_symbol}'
    os.makedirs(out_dir, exist_ok=True)

    if 'historical' in data and not data['historical'].empty:
        data['historical'].to_csv(f"{out_dir}/{actual_symbol}_historical.csv")

    if 'financials' in data and not data['financials'].empty:
        data['financials'].to_csv(f"{out_dir}/{actual_symbol}_financials.csv")

    if 'balance_sheet' in data and not data['balance_sheet'].empty:
        data['balance_sheet'].to_csv(f"{out_dir}/{actual_symbol}_balance_sheet.csv")

    if 'cashflow' in data and not data['cashflow'].empty:
        data['cashflow'].to_csv(f"{out_dir}/{actual_symbol}_cashflow.csv")

    if 'info' in data and data['info']:
        info_df = pd.DataFrame(list(data['info'].items()), columns=['Metric', 'Value'])
        info_df.to_csv(f"{out_dir}/{actual_symbol}_info.csv", index=False)
```

```python
def process_csv_file(csv_file, output_folder="data"):
    print(f"Processing: {csv_file}")
    print("-" * 50)
    
    companies = load_companies_from_csv(csv_file)
    print(f"Loaded {len(companies)} companies")
    

    successful = []
    failed = []
    
    for i, company in enumerate(companies, 1):
        symbol = company['Symbol']
        name = company['Company_Name']
        
        print(f"[{i}/{len(companies)}] {symbol} - {name}")

        data = fetch_data(symbol)
        
        if data is not None:
            save_data(data, symbol, output_folder)
            successful.append(symbol)
        else:
            failed.append(symbol)
        

        time.sleep(5)

    print(f"\nSUMMARY for {csv_file}:")
    print(f"Successful: {len(successful)}")
    print(f"Failed: {len(failed)}")
```

```python
process_csv_file('large_cap.csv', 'data/large_cap')

print("\n" + "=" * 40)

process_csv_file('mid_cap.csv', 'data/mid_cap')

print("\n" + "=" * 40)

process_csv_file('small_cap.csv', 'data/small_cap')
```
